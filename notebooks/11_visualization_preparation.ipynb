# 11 Visualization Preparation

## Purpose

This notebook prepares the final analytical datasets for visualization and presentation.

No new statistical tests, predictive models, or strategic classifications are introduced. Instead, the notebook consolidates the validated outputs from the previous stages into a small set of visualization-ready tables.

The final visual analysis is designed to answer the project's central business question:

**How competitive is Germany's tourism sector compared with other EU destinations, what factors explain its current position, and where are the strongest opportunities for future tourism growth?**

## Visualization Principles

The final dashboards and presentation should:

- distinguish tourism scale from tourism depth
- emphasize insight rather than display every available metric
- maintain consistent definitions across charts
- clearly distinguish actual data from projections
- identify source-market opportunities without implying causality
- treat counterfactual scenarios as illustrative benchmarks
- preserve data-quality and comparability limitations
- connect every visualization to a business implication

The final Tableau story will therefore follow the analytical sequence:

**Scale → Tourism Depth → Source Markets → Current Momentum → Outlook → Strategy**

# Tableau Visualization Architecture

The final visualization is designed as a business story rather than a collection of independent charts.

## Story Point 1: Germany Is a Major EU Tourism Destination

**Business question:** How strong is Germany's international tourism position?

**Primary dataset:** `tableau_eu_competitive_position.csv`

**Primary visual:** EU27 competitive-position scatter plot

- X-axis: Foreign Arrivals
- Y-axis: Foreign Nights
- Detail: Country
- Highlight: Germany
- Tooltip: Arrivals Rank, Nights Rank, Average Length of Stay

**Key message:** Germany has substantial international tourism scale, ranking fourth in arrivals and seventh in overnight stays in the complete 2024 EU27 comparison.

---

## Story Point 2: The Competitive Gap Is Tourism Depth

**Business question:** Why does Germany's strong visitor scale not translate into a higher overnight-stay position?

**Primary datasets:**
- `tableau_eu_los_2024.csv`
- `tableau_los_scenarios_2024.csv`

**Primary visuals:**
- EU27 average-length-of-stay ranking
- Illustrative LOS benchmark scenarios

**Key message:** Germany's average international stay is approximately 2.27 nights, ranking 21st among the EU27 in 2024.

The EU27 median scenario illustrates that even a moderate increase in stay duration could correspond to materially greater overnight-stay volume at Germany's existing arrival scale.

**Important:** Counterfactual scenarios are illustrative benchmarks, not forecasts or estimates of recoverable demand.

---

## Story Point 3: Different Markets Require Different Strategies

**Business question:** Where are Germany's strongest source-market opportunities?

**Primary dataset:** `tableau_source_market_strategy_2024.csv`

**Primary visual:** Source-market opportunity scatter plot

- X-axis: 2023–2024 Growth %
- Y-axis: Average Length of Stay
- Size: Foreign Nights
- Detail: Source Market
- Strategic segmentation: Strategic Role

**Key message:** Germany should not use a uniform international acquisition strategy. Established growth, retention, stay-extension, emerging-growth and long-stay markets require different objectives.

---

## Story Point 4: 2026 Growth Is Not Translating Fully into Tourism Depth

**Business question:** Is Germany's current tourism performance strengthening?

**Primary dataset:** `tableau_2026_ytd_growth.csv`

**Primary visual:** Monthly year-over-year percentage-change comparison

- Month: January–June
- Measure: Change %
- Series: Foreign Arrivals, Foreign Nights, Average Length of Stay

**Key message:** International arrivals increased throughout the first half of 2026, but overnight stays grew more slowly and average length of stay declined.

---

## Story Point 5: Near-Term Overnight-Stay Growth Remains Modest

**Business question:** What does the validated baseline indicate for the remainder of 2026?

**Primary datasets:**
- `tableau_2026_outlook.csv`
- `tableau_economic_context.csv`

**Primary visuals:**
- 2026 observed versus seasonal-naive baseline line chart
- Travel receipts and expenditure context

**Key message:** The seasonal-naive baseline suggests limited full-year overnight-stay growth. Inbound travel receipts have recovered strongly, but economic and tourism indicators should be interpreted within their separate statistical scopes.

**Important:** January–June values are observed. July–December values are seasonal-naive projections.

---

## Story Point 6: Shift from Volume-Only Growth to Tourism Depth

**Business question:** What should tourism decision-makers prioritize?

**Primary dataset:** `tableau_strategic_recommendations.csv`

**Primary visual:** Strategic recommendation and KPI summary

**Key message:** Germany should protect its strong international visitor base while increasing tourism depth through stay extension, differentiated source-market strategies, nights-based performance evaluation and continuous seasonal monitoring.

## Final Story

**Germany's tourism challenge is not primarily attracting international visitors. It is converting an already strong visitor position into greater overnight-stay depth and tourism value.**
